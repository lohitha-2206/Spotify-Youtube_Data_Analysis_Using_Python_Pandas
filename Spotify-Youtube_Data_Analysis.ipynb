## 1) Top 10 Artists - with the Highest Views on YouTube?

```python
data.head(2)
```

```python
Artist_grouped = data.groupby('Artist')['Views'].sum()
```

```python
Artist_grouped
```

```python
Artist_sorted = Artist_grouped.sort_values(ascending = False)
```

```python
Artist_sorted.head(10)
```

## 3) What are the most common Album Types on Spotify? How many tracks belong to each album type?A

## 4) How do the Average Views, Likes, and Comments are compared between different Album Types?

```python
data.head(1)
```

```python
df = data.groupby('Album_type')[['Likes', 'Views', 'Comments']].mean()
```

```python
df
```

```python
type(df)
```

```python
df = df.reset_index()
```

```python
df
```

```python
df_melted = pd.melt(df, id_vars = 'Album_type', var_name = 'Attribute', value_name = 'Total')
```

```python
df_melted
```

```python
plt.figure(figsize = (9,4) )
sns.barplot(x='Album_type', y='Total', hue='Attribute', data=df_melted)
```
